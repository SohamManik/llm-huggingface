```python
from transformers import pipeline

model = pipeline("text-generation")

def predict(prompt):
  completion = model(prompt)[0]["generated_text"]
  return completion
```

```python
predict("i loved her so much that i used to")
```

```python
gr.Interface(fn=predict, inputs="text", outputs="text").launch()
```

```python
import numpy as np

def reverse_audio(audio):
  sr , data = audio
  reversed_audio = (sr , np.flipud(data))
  return reversed_audio
```

```python
mic = gr.Audio(sources = ["microphone"] , type = "numpy" , label = "Speak Child..")

gr.Interface(fn = reverse_audio,inputs = mic , outputs = "audio").launch()
```

```python
model_2 = pipeline("automatic-speech-recognition")

def transcribe_audio(audio):
    transcription = model_2(audio)["text"]
    return transcription

gr.Interface(
    fn = transcribe_audio
    ,inputs = gr.Audio(type = "filepath"),
    outputs = "text"

    ).launch()
```
